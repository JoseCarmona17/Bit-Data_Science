```python
import pandas as pd
import numpy as np

df = pd.read_csv('imdb_top_movies.csv')
df.head()
```

Pregunta de análisis

¿Cuál es la duración promedio de las películas por década y qué década tiene las películas más largas?

```python
# Limpieza de datos
def convert_duration(d):
    try:
        h, m = 0, 0
        if "h" in d:
            h = int(d.split("h")[0])
            m = int(d.split("h")[1].replace("m", "").strip())
        return h*60 + m
    except:
        return np.nan

df["Duration_min"] = df["Duration"].apply(convert_duration)
df["Year"] = pd.to_numeric(df["Year"], errors="coerce")
```

```python
# Eliminar filas con datos esenciales faltantes
df = df.dropna(subset=["Duration_min", "Year"])
```

```python
# Crear columna Década
df["Decade"] = (df["Year"] // 10) * 10
```

```python
# Agrupación: duración promedio por década
decade_avg = df.groupby("Decade")["Duration_min"].mean().reset_index()
```

```python
# NumPy: array de duraciones
duration_array = df["Duration_min"].values

duration_mean = np.mean(duration_array)
duration_std = np.std(duration_array)
```

```python
# Normalización tipo z-score
duration_normalized = (duration_array - duration_mean) / duration_std

decade_avg.head(), duration_mean, duration_std
```

¿Cuál es la duración promedio de las películas por década y qué década tiene las películas más largas?

Las películas tienden a aumentar su duración a medida que avanzan las décadas.

De acuerdo con los resultados:

La década con las películas más largas es la de 1960, con un promedio de 146.37 minutos.
Esto quiere decir que en los años 60 hubo una tendencia hacia películas más extensas, posiblemente debido a cambios creativos y narrativos en la industria cinematográfica de esa época.
